```python
import os
import subprocess
import pandas as pd
import time

# Paths
csv_path = "/content/aptos2019/train.csv"
download_dir = "/content/aptos2019/train_images"

os.makedirs(download_dir, exist_ok=True)

# Read image IDs
train_df = pd.read_csv(csv_path)
image_ids = train_df["id_code"].tolist()

total = len(image_ids)
completed = 0
failed = []

print(f"Total training images: {total}")
print("Starting resumable download...")
print("-" * 50)

for i, image_id in enumerate(image_ids, start=1):

    png_path = os.path.join(download_dir, f"{image_id}.png")
    zip_path = os.path.join("/content/aptos2019", f"{image_id}.png.zip")

    # Skip images already downloaded
    if os.path.exists(png_path):
        completed += 1
        continue

    try:
        result = subprocess.run(
            [
                "kaggle",
                "competitions",
                "download",
                "-c",
                "aptos2019-blindness-detection",
                "-f",
                f"train_images/{image_id}.png",
                "-p",
                "/content/aptos2019"
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.PIPE,
            text=True
        )

        if result.returncode != 0:
            failed.append(image_id)
            print(f"❌ Failed: {image_id}")
            continue

        # Extract the downloaded ZIP
        result = subprocess.run(
            [
                "unzip",
                "-o",
                zip_path,
                "-d",
                download_dir
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.PIPE,
            text=True
        )

        if result.returncode != 0:
            failed.append(image_id)
            print(f"❌ Extraction failed: {image_id}")
            continue

        # Remove temporary ZIP
        if os.path.exists(zip_path):
            os.remove(zip_path)

        completed += 1

    except Exception as e:
        failed.append(image_id)
        print(f"❌ Error: {image_id} -> {e}")

    # Progress update
    if i % 25 == 0 or i == total:
        print(
            f"Progress: {i}/{total} | "
            f"Downloaded: {completed} | "
            f"Failed: {len(failed)}"
        )

print("-" * 50)
print("Download process finished.")
print(f"Successfully available: {completed}")
print(f"Failed: {len(failed)}")

if failed:
    print("\nFailed image IDs:")
    print(failed[:20])
```

```python
import os
import subprocess
import pandas as pd
import time

# Paths
csv_path = "/content/aptos2019/train.csv"
download_dir = "/content/aptos2019/train_images"
temp_dir = "/content/aptos2019"

os.makedirs(download_dir, exist_ok=True)

# Load image IDs
train_df = pd.read_csv(csv_path)
image_ids = train_df["id_code"].tolist()

total = len(image_ids)

print(f"Total training images: {total}")
print(f"Already downloaded: {len(os.listdir(download_dir))}")
print("The downloader will resume from existing images.")
print("-" * 60)

failed = []

for i, image_id in enumerate(image_ids, start=1):

    png_path = os.path.join(download_dir, f"{image_id}.png")
    zip_path = os.path.join(temp_dir, f"{image_id}.png.zip")

    # Already successfully downloaded
    if os.path.exists(png_path):
        continue

    # Remove an old incomplete ZIP if one exists
    if os.path.exists(zip_path):
        os.remove(zip_path)

    try:
        # Download individual Kaggle file
        result = subprocess.run(
            [
                "kaggle",
                "competitions",
                "download",
                "-c",
                "aptos2019-blindness-detection",
                "-f",
                f"train_images/{image_id}.png",
                "-p",
                temp_dir
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.PIPE,
            text=True,
            timeout=120
        )

        # Check Kaggle command itself
        if result.returncode != 0:
            failed.append(image_id)
            print(f"❌ Download failed: {image_id}")
            continue

        # IMPORTANT: verify ZIP actually exists
        if not os.path.exists(zip_path):
            failed.append(image_id)
            print(f"❌ ZIP not created: {image_id}")
            continue

        # Extract ZIP
        result = subprocess.run(
            [
                "unzip",
                "-o",
                zip_path,
                "-d",
                download_dir
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.PIPE,
            text=True
        )

        # Verify extracted PNG exists
        if result.returncode != 0 or not os.path.exists(png_path):
            failed.append(image_id)
            print(f"❌ Extraction failed: {image_id}")
            continue

        # Only delete ZIP after successful extraction
        os.remove(zip_path)

    except subprocess.TimeoutExpired:
        failed.append(image_id)
        print(f"⏱️ Timeout: {image_id}")

    except Exception as e:
        failed.append(image_id)
        print(f"❌ Error: {image_id} -> {e}")

    # Progress every 25 images
    if i % 25 == 0:
        current = len(os.listdir(download_dir))
        print(
            f"Progress: {i}/{total} | "
            f"Images available: {current} | "
            f"Failures this run: {len(failed)}"
        )

print("-" * 60)
print("Downloader stopped/finished.")
print(f"Images currently available: {len(os.listdir(download_dir))}")
print(f"Failures this run: {len(failed)}")

if failed:
    print("\nFirst 20 failed IDs:")
    print(failed[:20])
```

```python
import os
import shutil
import subprocess
import pandas as pd

# --------------------------------------------------
# Paths
# --------------------------------------------------

csv_path = "/content/aptos2019/train.csv"
temp_dir = "/content/aptos2019"
download_dir = "/content/aptos2019/train_images"

os.makedirs(download_dir, exist_ok=True)

# --------------------------------------------------
# Load image IDs
# --------------------------------------------------

train_df = pd.read_csv(csv_path)
image_ids = train_df["id_code"].astype(str).tolist()

total = len(image_ids)

print(f"Total training images: {total}")
print(f"Images already in training folder: {len(os.listdir(download_dir))}")
print("-" * 60)

# --------------------------------------------------
# Download remaining images
# --------------------------------------------------

failed = []

for i, image_id in enumerate(image_ids, start=1):

    final_path = os.path.join(download_dir, f"{image_id}.png")
    direct_path = os.path.join(temp_dir, f"{image_id}.png")
    zip_path = os.path.join(temp_dir, f"{image_id}.png.zip")

    # Already downloaded
    if os.path.exists(final_path):
        continue

    try:

        # Remove stale temporary files
        if os.path.exists(direct_path):
            os.remove(direct_path)

        if os.path.exists(zip_path):
            os.remove(zip_path)

        # Ask Kaggle for the image
        result = subprocess.run(
            [
                "kaggle",
                "competitions",
                "download",
                "-c",
                "aptos2019-blindness-detection",
                "-f",
                f"train_images/{image_id}.png",
                "-p",
                temp_dir,
                "-o"
            ],
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True,
            timeout=120
        )

        # --------------------------------------------------
        # Case 1: Kaggle downloaded PNG directly
        # --------------------------------------------------

        if os.path.exists(direct_path):

            shutil.move(direct_path, final_path)

        # --------------------------------------------------
        # Case 2: Kaggle downloaded ZIP
        # --------------------------------------------------

        elif os.path.exists(zip_path):

            extract_result = subprocess.run(
                [
                    "unzip",
                    "-o",
                    zip_path,
                    "-d",
                    download_dir
                ],
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True
            )

            if extract_result.returncode != 0 or not os.path.exists(final_path):
                failed.append(image_id)
                print(f"❌ Extraction failed: {image_id}")
                continue

            os.remove(zip_path)

        # --------------------------------------------------
        # Case 3: Nothing was downloaded
        # --------------------------------------------------

        else:

            failed.append(image_id)
            print(f"❌ Download failed: {image_id}")

            # Show Kaggle's message for diagnosis
            if result.stderr:
                print("   ", result.stderr.strip()[:200])

            continue

    except subprocess.TimeoutExpired:

        failed.append(image_id)
        print(f"⏱️ Timeout: {image_id}")

    except Exception as e:

        failed.append(image_id)
        print(f"❌ Error: {image_id} -> {e}")

    # --------------------------------------------------
    # Progress
    # --------------------------------------------------

    if i % 25 == 0 or i == total:

        current = len(os.listdir(download_dir))

        print(
            f"Progress: {i}/{total} | "
            f"Images available: {current} | "
            f"Failures: {len(failed)}"
        )

print("-" * 60)
print("Download process finished/stopped.")
print(f"Images currently available: {len(os.listdir(download_dir))}")
print(f"Failures during this run: {len(failed)}")

if failed:
    print("\nFirst 20 failed IDs:")
    print(failed[:20])
```

```python
import os
import pandas as pd

train_df = pd.read_csv("/content/aptos2019/train.csv")
image_dir = "/content/aptos2019/train_images"

downloaded_ids = {
    os.path.splitext(filename)[0]
    for filename in os.listdir(image_dir)
    if filename.lower().endswith(".png")
}

labeled_ids = set(train_df["id_code"].astype(str))

missing_labels = downloaded_ids - labeled_ids

print("Images downloaded:", len(downloaded_ids))
print("Images with matching labels:", len(downloaded_ids & labeled_ids))
print("Downloaded images without a label:", len(missing_labels))

if missing_labels:
    print("Unexpected IDs:", list(missing_labels)[:10])
```

```python
import os
import json
import pandas as pd
from datetime import datetime

image_dir = "/content/aptos2019/train_images"
results_dir = "/content/drive/MyDrive/DR-Visualize/results"

os.makedirs(results_dir, exist_ok=True)

downloaded_ids = sorted([
    os.path.splitext(f)[0]
    for f in os.listdir(image_dir)
    if f.lower().endswith(".png")
])

progress = {
    "dataset": "APTOS 2019 Blindness Detection",
    "total_labeled_images": 3662,
    "downloaded_images": len(downloaded_ids),
    "remaining_images": 3662 - len(downloaded_ids),
    "last_updated": datetime.now().isoformat(),
    "downloaded_ids": downloaded_ids
}

progress_path = os.path.join(
    results_dir,
    "aptos_download_progress.json"
)

with open(progress_path, "w") as f:
    json.dump(progress, f, indent=2)

print("Progress saved to:", progress_path)
print("Downloaded:", len(downloaded_ids))
print("Remaining:", 3662 - len(downloaded_ids))
```
